```python

import time
import numpy as np
import pandas as pd
from collections import Counter

from sklearn.datasets import load_iris, load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, confusion_matrix
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC







```

```python
### Class kNN Classifier: Definies custom Knn

class KNNClassifier:
    def __init__(self, k=3):
        self.k = k

    def fit(self, X, y):
        self.X_train = X
        self.y_train = y

    def _euclidean_distance(self, a, b):
        return np.sqrt(np.sum((a - b) ** 2))
### predicts more than one point at a time as X is a list
    def predict(self, X):
        predictions = []

        for x in X:
            distances = []

            for xi, yi in zip(self.X_train, self.y_train):
                d = self._euclidean_distance(x, xi)
                distances.append((d, yi))

            distances.sort(key=lambda x: x[0])
            k_nearest_neighbors = distances[:self.k]
            labels = []
            for pair in k_nearest_neighbors:
                label = pair[1]
                labels.append(label)


            prediction = max(set(labels), key=labels.count)

            predictions.append(prediction)




        return np.array(predictions)
```

```python
## Evaluation function: mainly uses the methods  from Sklitlearn to find the time takes
## Predcit and train and also verifies the accuracy and also gives the confusion matrix
## of the model.
def evaluate_model(model, X_train, X_test, y_train, y_test):
    # Training time
    start_train = time.time()
    model.fit(X_train, y_train)
    training_time = time.time() - start_train

    # Prediction time
    start_test = time.time()
    y_pred = model.predict(X_test)
    testing_time = time.time() - start_test

    accuracy = accuracy_score(y_test, y_pred)
    cm = confusion_matrix(y_test, y_pred)

    return accuracy, training_time, testing_time, cm
```

```python
### Loading the two data sets given in the project (Optional) and parsing it in
##  labels and attrubute values
##  First two data sets given under the project tasks.
def load_dataset(name):
    if name == "iris":
        data = load_iris()
    elif name == "breast_cancer":
        data = load_breast_cancer()
    else:
        raise ValueError("Unknown dataset")

    return data.data, data.target
```

```python
## To produce the test samples to test and to calculate the parameters such as accuracy
## confussion matrix and prediction time

def run_experiment(dataset_name):
    X, y = load_dataset(dataset_name)

    X_train, X_test, y_train, y_test = train_test_split(
        X, y, test_size=0.3, stratify=y
    )


    # Definied Knn
    custom_knn = KNNClassifier(k=5)
    custom_results = evaluate_model(
        custom_knn, X_train, X_test, y_train, y_test
    )

    # Direct KNN from the Skitlearn
    sklearn_knn = KNeighborsClassifier(n_neighbors=5)
    sklearn_knn_results = evaluate_model(
        sklearn_knn, X_train, X_test, y_train, y_test
    )

    ###Third Classifier from skitlearn namely Support Vector Machines
    svm = SVC(kernel="rbf")
    svm_results = evaluate_model(
        svm, X_train, X_test, y_train, y_test
    )

    return {
        "definied_KNN": custom_results,
        "Sklearn KNN": sklearn_knn_results,
        "SVM Results": svm_results
    }
```
